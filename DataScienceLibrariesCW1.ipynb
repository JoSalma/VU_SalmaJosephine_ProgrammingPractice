# 1 — NumPy

## What is NumPy?

**NumPy** stands for **Numerical Python**.

It is designed for efficient numerical computing and stores data in multidimensional objects called **ndarrays**.

NumPy is useful for:
- arrays and matrices,
- vectorized calculations,
- descriptive statistics,
- filtering,
- missing numeric values,
- and fast aggregation.


## 1.1 Creating a One-Dimensional Array

A one-dimensional NumPy array is similar to a Python list, but it is optimized for numerical calculations.

### Syntax
`np.array(values)`


```python
# Create a one-dimensional NumPy array
marks = np.array([65, 72, 48, 81, 55])

print("Marks:", marks)
print("Type:", type(marks))
```

### Interpretation

`marks` is now a NumPy `ndarray`.

Unlike an ordinary Python list, NumPy can apply numerical operations directly to all values in the array.


## 1.2 Creating Arrays Automatically

NumPy can generate values instead of requiring us to type every value.

- `np.arange()` → sequence using a start and stop
- `np.linspace()` → evenly spaced values
- `np.zeros()` → values initialized to zero
- `np.ones()` → values initialized to one


```python
a = np.arange(1, 11)
b = np.linspace(0, 100, 5)
zeros = np.zeros(5)
ones = np.ones((3, 2))

print("arange:", a)
print("linspace:", b)
print("zeros:", zeros)
print("ones:\n", ones)
```

### Interpretation

`np.arange(1, 11)` stops **before 11**, so it returns 1 to 10.

`np.linspace(0, 100, 5)` includes both endpoints and divides the interval into five evenly spaced values.


## 1.3 Arrays and Matrices

A matrix is represented as a two-dimensional NumPy array.

Important properties:
- `shape` → rows and columns
- `ndim` → number of dimensions
- `size` → total number of values
- `dtype` → stored data type


```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    
])

print(matrix)
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

### Interpretation

The matrix has:
- **2 rows**
- **3 columns**
- **2 dimensions**
- **6 total values**

Therefore, its shape is `(2, 3)`.


## 1.4 Indexing and Slicing

Python uses **zero-based indexing**.

This means:
- first position → index `0`
- second position → index `1`

For a matrix, we can select a row, column, or individual element.


```python
print("First row:", matrix[0])
print("Second row:", matrix[1])
print("First column:", matrix[:, 0])
print ("second column:", matrix [:, 1])
print("Row 1, Column 2 value:", matrix[0, 1])
```

## 1.5 Mathematical Operations and Vectorization

**Vectorization** means performing an operation on an entire array without writing a loop for each element.


```python
prices = np.array([850, 500, 300, 850, 650])
quantities = np.array([2, 1, 4, 1, 3])

# Element-by-element multiplication
revenue = prices * quantities

print("Prices:", prices)
print("Quantities:", quantities)
print("Revenue:", revenue)
```

### Interpretation

Each price is multiplied by the quantity at the same position.

For example:

`850 × 2 = 1700`

NumPy performs all five calculations in one vectorized statement.


## 1.6 Aggregation

Aggregation summarizes many values into useful statistics.

Common functions:
- `np.sum()`
- `np.mean()`
- `np.median()`
- `np.min()`
- `np.max()`


```python
print("Total revenue:", np.sum(revenue))
print("Average price:", np.mean(prices))
print("Median price:", np.median(prices))
print("Lowest price:", np.min(prices))
print("Highest price:", np.max(prices))
```

## 1.7 Missing Numeric Values

NumPy uses `np.nan` to represent a missing numerical value.

Useful functions include:
- `np.isnan()` → detect missing values
- `np.nanmean()` → mean ignoring NaN
- `np.nanmedian()` → median ignoring NaN
- `np.nansum()` → sum ignoring NaN


```python
quantity = np.array([2, 1, np.nan, 1, 3, 2, 4])

missing = np.isnan(quantity)
median_qty = np.nanmedian(quantity)

print("Original:", quantity)
print("Missing mask:", missing)
print("Number missing:", missing.sum())
print("Median ignoring NaN:", median_qty)
```

## 1.8 Replacing Missing Values with `np.where()`

`np.where(condition, value_if_true, value_if_false)` applies a conditional replacement.


```python
quantity_clean = np.where(
    missing,
    median_qty,
    quantity
)

print("Before:", quantity)
print("After :", quantity_clean)
```

### Why use the median?

The median is often useful when extreme values or outliers could distort the mean.

The correct cleaning method always depends on the meaning of the data.


## 1.9 Boolean Filtering

A **Boolean mask** contains `True` and `False`.

When the mask is applied to an array, NumPy keeps only positions where the condition is `True`.


```python
mask = prices > 500
high_prices = prices[mask]

print("Mask:", mask)
print("Prices above 500:", high_prices)

# Multiple conditions: use & for AND
products = np.array(["Laptop", "Phone", "Tablet", "Laptop", "Monitor"])
laptop_mask = (products == "Laptop") & (quantities >= 1)

print("Matching products:", products[laptop_mask])
```

## 1.10 Combining Arrays

Two useful methods are:

- `np.concatenate()` → joins values end-to-end
- `np.column_stack()` → places arrays beside each other as columns


```python
jan_sales = np.array([100, 150, 120])
feb_sales = np.array([130, 170, 160])

all_sales = np.concatenate([jan_sales, feb_sales])
sales_table = np.column_stack([jan_sales, feb_sales])

print("Concatenated:", all_sales)
print("\nColumn-stacked table:\n", sales_table)
```

# 5 — Scikit-learn

## What is Scikit-learn?

**Scikit-learn** is a Python machine-learning library.

We continue using the same cleaned **Wine Quality dataset (`df`)**.

Our task is **classification** because the model will predict the discrete wine `quality` score.

The workflow is:

**Choose Features and Target → Split Data → Train Model → Predict → Evaluate**


## 5.1 Define the Features (`X`) and Target (`y`)

Instead of selecting columns automatically, we explicitly list the exact predictor columns used by the model.

### Features (`X`)
The model uses these 11 original numerical wine measurements:

1. `fixed_acidity`
2. `volatile_acidity`
3. `citric_acid`
4. `residual_sugar`
5. `chlorides`
6. `free_sulfur_dioxide`
7. `total_sulfur_dioxide`
8. `density`
9. `ph`
10. `sulphates`
11. `alcohol`

### Target (`y`)
- `quality`

`wine_type`, `quality_band`, and `alcohol_band` are not included as model features. The calculated Pandas columns are also excluded so that the model uses the original numerical wine measurements.


```python
X = df[[
    "fixed_acidity",
    "volatile_acidity",
    "citric_acid",
    "residual_sugar",
    "chlorides",
    "free_sulfur_dioxide",
    "total_sulfur_dioxide",
    "density",
    "ph",
    "sulphates",
    "alcohol"
]]

y = df["quality"]

X.head()
```

```python
y.head()
```

## 5.2 Confirm the Machine-Learning Data

Before splitting the dataset, confirm the number of records and features and check the quality classes.


```python
print("X shape:", X.shape)
print("y shape:", y.shape)

y.value_counts().sort_index()
```

## 5.3 Split into Training and Testing Data

`train_test_split()` divides the data into:

- **80% training data** — used to teach the model.
- **20% testing data** — used to evaluate the trained model.

`random_state=42` makes the split reproducible.

`stratify=y` helps preserve the distribution of the quality classes in both sets.


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

## 5.4 Train a Decision Tree Classifier

A **Decision Tree Classifier** learns decision rules from the training data.

`X_train` contains the 11 wine measurements and `y_train` contains their known quality scores.


```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

model.fit(X_train, y_train)
```

## 5.5 Make Predictions

The trained model now predicts wine quality for the unseen records in `X_test`.

The predictions are stored in `y_pred`.


```python
y_pred = model.predict(X_test)

y_pred[:10]
```

## 5.6 Compare Actual and Predicted Quality

This table places the true quality score beside the model's predicted quality score.


```python
comparison = pd.DataFrame({
    "Actual_Quality": y_test.to_numpy(),
    "Predicted_Quality": y_pred
})

comparison.head(10)
```

## 5.7 Evaluate the Model

We use:

- **Accuracy** — proportion of all predictions that are correct.
- **Precision** — how many predictions for a class are correct.
- **Recall** — how many actual records of a class are correctly found.
- **F1-score** — balance between precision and recall.


```python
from sklearn.metrics import accuracy_score, classification_report

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy, 3))
print()
print(classification_report(y_test, y_pred, zero_division=0))
```

## 5.8 Confusion Matrix

A confusion matrix compares actual quality classes with predicted quality classes.

Correct predictions appear on the main diagonal. Values outside the diagonal show misclassifications.


```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred
)

plt.title("Wine Quality Confusion Matrix")
plt.show()
```

## 5.9 Feature Importance

A Decision Tree can estimate the relative contribution of each input feature to its decisions.


```python
feature_importance = pd.Series(
    model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

feature_importance
```

## 5.10 Visualize Feature Importance

The bar chart makes it easier to compare which wine measurements contributed more to the Decision Tree's decisions.

# 6 — Statsmodels

## What is Statsmodels?

**Statsmodels** is a Python library for statistical modelling and statistical inference.

Unlike the Scikit-learn section, which focuses mainly on prediction, this section focuses on **explaining relationships between variables**.

We continue using the same cleaned **Wine Quality DataFrame `df`**.

In this practical we use **Ordinary Least Squares (OLS) regression** to study whether wine measurements help explain the numerical `quality` score.

> If Statsmodels is not installed, run `pip install statsmodels` in your terminal or Anaconda Prompt.


## 6.1 Import Statsmodels

`statsmodels.api` is commonly imported as `sm`.

We use it to create and fit the OLS regression model.


```python
import statsmodels.api as sm

print("Statsmodels imported successfully.")
```

## 6.2 Define the Statistical Model Variables

For a simple and easy-to-explain model:

- `X_stats` contains the independent variables or predictors.
- `y_stats` contains the dependent variable we want to explain.

We use:

- `alcohol`
- `volatile_acidity`
- `sulphates`

to explain the numerical wine `quality` score.


```python
X_stats = df[[
    "alcohol",
    "volatile_acidity",
    "sulphates"
]]

y_stats = df["quality"]

X_stats.head()
```

## 6.3 Add the Constant

A linear regression equation normally contains an **intercept**.

`sm.add_constant()` adds a column of `1`s so Statsmodels can estimate that intercept.

After this step, `const` represents the intercept term.


```python
X_stats = sm.add_constant(X_stats)

X_stats.head()
```

## 6.4 Fit the OLS Regression Model

`sm.OLS(y, X)` creates an **Ordinary Least Squares** regression model.

`.fit()` estimates the regression coefficients.

OLS chooses coefficients that minimize the squared differences between the actual and predicted values.


```python
ols_model = sm.OLS(y_stats, X_stats).fit()

print("OLS model fitted successfully.")
```

## 6.5 Display the Statistical Summary

`summary()` provides a detailed statistical report.

Important parts to explain:

- **R-squared** — proportion of variation in `quality` explained by the predictors.
- **coef** — estimated effect of each predictor.
- **P>|t|** — p-value used to assess statistical significance.
- **const** — estimated intercept.
- **F-statistic** — tests the overall regression model.

A commonly used significance level is **0.05**. A p-value below 0.05 is often treated as evidence that the predictor has a statistically significant relationship with the dependent variable, subject to the model assumptions.


```python
print(ols_model.summary())
```

## 6.6 View the Regression Coefficients

The coefficients are easier to inspect separately using `.params`.

The sign of a coefficient shows the direction of the relationship:

- positive coefficient → predicted quality increases as that predictor increases, holding the other predictors constant;
- negative coefficient → predicted quality decreases as that predictor increases, holding the other predictors constant.


```python
ols_model.params
```

## 6.7 View the P-values

`.pvalues` returns the p-value for each coefficient.

This helps us examine which variables provide stronger statistical evidence of association with wine quality in this fitted model.


```python
ols_model.pvalues
```

## 6.8 R-squared and Adjusted R-squared

`rsquared` measures the proportion of variation explained by the model.

`rsquared_adj` adjusts R-squared for the number of predictors used.


```python
print("R-squared:", round(ols_model.rsquared, 3))
print("Adjusted R-squared:", round(ols_model.rsquared_adj, 3))
```

## 6.9 Actual and Fitted Quality Values

`fittedvalues` contains the quality values estimated by the regression model.

We compare a few actual values with their fitted values.


```python
stats_comparison = pd.DataFrame({
    "Actual_Quality": y_stats,
    "Fitted_Quality": ols_model.fittedvalues
})

stats_comparison.head(10)
```

## 6.10 Visualize Actual vs Fitted Quality

This scatter plot compares the observed quality scores with the values fitted by the OLS model.

A perfect model would place all observations on the diagonal reference line.


```python
plt.figure(figsize=(8, 5))
plt.scatter(
    stats_comparison["Actual_Quality"],
    stats_comparison["Fitted_Quality"],
    alpha=0.4
)

minimum = min(
    stats_comparison["Actual_Quality"].min(),
    stats_comparison["Fitted_Quality"].min()
)

maximum = max(
    stats_comparison["Actual_Quality"].max(),
    stats_comparison["Fitted_Quality"].max()
)

plt.plot([minimum, maximum], [minimum, maximum])
plt.title("Statsmodels OLS: Actual vs Fitted Wine Quality")
plt.xlabel("Actual Quality")
plt.ylabel("Fitted Quality")
plt.show()
```
